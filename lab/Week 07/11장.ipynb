
## 1. 기본 함수 (A~B)

| 함수 | 설명 | 예시 |
|---|---|---|
| `abs()` | 숫자의 절댓값을 구함 | `abs(-5)` → `5` |
| `all()` | 모든 요소가 참이면 `True` | `all([1, 2, 0])` → `False` |
| `any()` | 하나라도 참이면 `True` | `any([0, 0, 1])` → `True` |
| `ascii()` | 문자열 등을 ASCII 형태로 표현하며 한글 등은 이스케이프 처리 | `ascii("가")` → `"'\\uac00'"` |
| `bin()` | 정수를 2진수 문자열로 변환 | `bin(10)` → `'0b1010'` |
| `bool()` | 값을 참(`True`) 또는 거짓(`False`)으로 변환 | `bool(0)` → `False` |
| `breakpoint()` | 프로그램을 일시 중단하고 디버거 실행 | `breakpoint()` |
| `bytearray()` | 수정 가능한 바이트 데이터 생성 | `bytearray([65, 66])` |
| `bytes()` | 수정할 수 없는 바이트 데이터 생성 | `bytes([65, 66])` → `b'AB'` |

## 2. 객체 및 실행 함수 (D~F)

| 함수 | 설명 | 예시 |
|---|---|---|
| `delattr()` | 객체에 있는 속성을 삭제 | `delattr(obj, "name")` |
| `dict()` | 딕셔너리 생성 | `dict(a=1, b=2)` |
| `dir()` | 객체가 가진 속성과 메서드 목록 확인 | `dir(str)` |
| `divmod()` | 나눗셈의 몫과 나머지를 동시에 반환 | `divmod(10, 3)` → `(3, 1)` |
| `enumerate()` | 반복할 때 번호를 함께 붙여줌 | `list(enumerate(['a','b']))` |
| `eval()` | 문자열로 작성된 파이썬 표현식을 계산 | `eval("2+3")` → `5` |
| `exec()` | 문자열로 작성된 파이썬 코드를 실행 | `exec("print(10)")` |
| `filter()` | 조건에 맞는 요소만 걸러냄 | `filter(조건함수, 리스트)` |
| `float()` | 값을 실수로 변환 | `float("3.14")` → `3.14` |

> **주의:** `eval()`과 `exec()`는 외부에서 받은 문자열을 그대로 실행하면 위험할 수 있으므로 주의해야 합니다.

## 3. 자료형 및 확인 함수 (H~I)

| 함수 | 설명 | 예시 |
|---|---|---|
| `hash()` | 객체의 해시값(정수)을 구함 | `hash(10)` → `10` |
| `help()` | 함수나 객체의 사용법을 보여줌 | `help(print)` |
| `hex()` | 정수를 16진수 문자열로 변환 | `hex(255)` → `'0xff'` |
| `id()` | 객체의 고유 식별값을 반환 | `id(a)` |
| `input()` | 사용자에게 키보드 입력을 받음 | `input("이름: ")` |
| `int()` | 값을 정수로 변환 | `int("10")` → `10` |
| `isinstance()` | 객체가 특정 자료형이나 클래스에 속하는지 검사 | `isinstance(10, int)` → `True` |
| `issubclass()` | 클래스가 다른 클래스의 하위 클래스인지 확인 | `issubclass(bool, int)` → `True` |
| `iter()` | 반복할 수 있는 객체에서 반복자를 생성 | `iter([1, 2, 3])` |

## 4. 데이터 처리 함수 (M~P)

| 함수 | 설명 | 예시 |
|---|---|---|
| `memoryview()` | 바이트 데이터를 복사하지 않고 접근하는 객체 생성 | `memoryview(b"ABC")` |
| `min()` | 가장 작은 값을 반환 | `min(3, 1, 5)` → `1` |
| `next()` | 반복자에서 다음 요소를 가져옴 | `next(iter([10, 20]))` → `10` |
| `object()` | 파이썬의 가장 기본적인 객체를 생성 | `object()` |
| `oct()` | 정수를 8진수 문자열로 변환 | `oct(8)` → `'0o10'` |
| `open()` | 파일을 열어 읽거나 쓸 수 있게 함 | `open("a.txt", "r")` |
| `ord()` | 문자를 유니코드 숫자로 변환 | `ord("A")` → `65` |
| `pow()` | 거듭제곱을 계산 | `pow(2, 3)` → `8` |
| `print()` | 화면에 내용을 출력 | `print("Hello")` |

## 5. 집합, 문자열 및 기타 함수 (S~T)

| 함수 | 설명 | 예시 |
|---|---|---|
| `set()` | 중복을 허용하지 않는 집합 생성 | `set([1,1,2])` → `{1,2}` |
| `setattr()` | 객체에 속성을 추가하거나 값을 변경 | `setattr(obj, "age", 20)` |
| `slice()` | 슬라이싱할 범위를 지정하는 객체 생성 | `slice(1, 4)` |
| `sorted()` | 데이터를 정렬한 새 리스트 반환 | `sorted([3,1,2])` → `[1,2,3]` |
| `staticmethod()` | 클래스에서 정적 메서드를 정의할 때 사용 | `@staticmethod` |
| `str()` | 값을 문자열로 변환 | `str(123)` → `'123'` |
| `sum()` | 숫자들의 합계를 구함 | `sum([1,2,3])` → `6` |
| `super()` | 부모 클래스의 메서드 등을 사용할 때 쓰임 | `super().__init__()` |
| `tuple()` | 수정할 수 없는 튜플 생성 | `tuple([1,2])` → `(1,2)` |
