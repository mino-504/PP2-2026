## 리스트 비교 연산

```python
list1 = [1, 2, 3]
list2 = [1, 2, 3]
print(list1 == list2)
```

```python
list1 = [3, 4, 5]
list2 = [1, 2, 3]
print(list1 > list2)
```

| 연산자 | 의미               | 예시                     | 결과   |
| --- | ---------------- | ---------------------- | ---- |
| ==  | 두 리스트가 같은지 비교    | [1, 2, 3] == [1, 2, 3] | True |
| !=  | 두 리스트가 다른지 비교    | [1, 2, 3] != [1, 2, 4] | True |
| >   | 왼쪽 리스트가 더 큰지 비교  | [3, 4, 5] > [1, 2, 3]  | True |
| <   | 왼쪽 리스트가 더 작은지 비교 | [1, 2, 3] < [3, 4, 5]  | True |
| >=  | 크거나 같은지 비교       | [1, 2, 3] >= [1, 2, 3] | True |
| <=  | 작거나 같은지 비교       | [1, 2, 3] <= [1, 2, 3] | True |


>첫 번째 비교  
→ 같으면 두 번째  
→ 또 같으면 세 번째  
→ 처음으로 다른 값이 나온 위치에서 결정

# 리스트 복사 정리

| 방식 | 복사 형태 | 바깥 리스트 | 안쪽 mutable 객체 | 예시 | 결과 |
|---|---|---|---|---|---|
| values = temps | 복사 X | 공유 | 공유 | temps = [1, 2, 3]<br>values = temps<br>temps[0] = 99 | temps = [99, 2, 3]<br>values = [99, 2, 3] |
| values = list(temps) | 얕은 복사 | 복제 | 공유 | temps = [[1, 2], 3, 4]<br>values = list(temps)<br>temps[0][0] = 7 | temps = [[7, 2], 3, 4]<br>values = [[7, 2], 3, 4] |
| values = copy.deepcopy(temps) | 깊은 복사 | 복제 | 복제 | import copy<br>temps = [[1, 2], 3, 4]<br>values = copy.deepcopy(temps)<br>temps[0][0] = 7 | temps = [[7, 2], 3, 4]<br>values = [[1, 2], 3, 4] |

## 핵심

| 내용 | 설명 |
|---|---|
| values = temps | 새로운 리스트를 만들지 않고 같은 리스트 전체를 공유 |
| list(temps) | 항상 얕은 복사이며 바깥 리스트만 새로 만듦 |
| deepcopy(temps) | 안쪽 객체까지 모두 새로 복제 |
| 1차원 리스트 | int, str처럼 수정할 수 없는 값만 있으면 얕은 복사도 깊은 복사처럼 보일 수 있음 |
| 2중 리스트 | 안쪽 리스트를 공유하기 때문에 안쪽 값을 수정하면 복사본에도 영향을 줄 수 있음 |
| mutable 객체 | list, dict, set처럼 내부 내용을 수정할 수 있는 객체 |  

<br>

> 한 줄 정리  
> temps → 전체 공유  
> list(temps) → 겉만 복제, 안쪽 mutable 객체는 공유  
> copy.deepcopy(temps) → 안쪽까지 전부 복제
