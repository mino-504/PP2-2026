## Typehint 연습

이 파일은 파이썬의 typehint 예제와 이론을 실습한 결과입니다.

타입힌트는 함수의 매개변수와 반환값의 타입을 지정하는 목적으로 사용하며, 엄격한 타입지정으로 인한 프로그래머의 의도를 명확히 전달하기 위함이다.   
안써도 되지만 파이썬 13부터는 쓰는 것을 권장하고 있다.

```python
def greet(msg: str) -> str:
    print("in greet:", msg)
    return msg + "!"

print(greet("안녕"))
```

| 타입 힌트           | 의미                       | 예시 값                      | 활용 상황                |
| --------------- | ------------------------ | ------------------------- | -------------------- |
| Tuple[int, int] | 길이가 정확히 2이고, 두 요소 모두 int | (10, 20)                  | 좌표 (x, y) 표현, 두 값 반환 |
| Tuple[int, ...] | 길이가 가변적이고 모든 요소가 int     | (1, 2, 3, 4)              | 여러 정수 묶음 반환          |
| List[int]       | 정수형 요소들로 이루어진 리스트        | [1, 2, 3, 4]              | 반복 가능한 정수 집합, 동적 크기  |
| Dict[str, int]  | 키는 문자열, 값은 정수            | {"apple": 3, "banana": 5} | 매핑 구조, 데이터 카운팅       |
| Set[int]        | 중복 없는 정수 집합              | {1, 2, 3}                 | 유일한 값 모음, 집합 연산      |
| Optional[int]   | int 또는 None              | 42 또는 None                | 값이 없을 수도 있는 경우       |


```python
from typing import Tuple, List, Dict, Set, Optional

def example() -> None:
    point: Tuple[int, int] = (5, 7) # 좌표
    numbers: List[int] = [1, 2, 3, 4] # 정수 리스트
    fruit_counts: Dict[str, int] = {"apple": 3, "banana": 5}
    unique_ids: Set[int] = {101, 102, 103} # 중복 없는 ID 집합
    maybe_value: Optional[int] = None # 값이 없을 수도 있음
    print(point, numbers, fruit_counts, unique_ids, maybe_value)
    
example()
```

| 구분    | 대문자 (Tuple, List, Dict)                       | 소문자 (tuple, list, dict)   |
| ----- | --------------------------------------------- | ------------------------- |
| 제공 위치 | typing 모듈 (타입 힌트용)                            | 파이썬 내장 자료형                |
| 목적    | 정적 타입 힌트: 코드 검사기(mypy, pyright)가 타입을 확인할 때 사용 | 실제 데이터 구조: 프로그램 실행 시 값 저장 |
| 예시    | Tuple[int, int] → 정수 2개짜리 튜플                  | (10, 20) → 실제 튜플 객체       |
| 특징    | 타입을 설명하는 설명서 역할                               | 데이터를 담는 그릇 역할             |
| 검사 시점 | 정적 검사기가 확인 (런타임 강제 X)                         | 런타임에서 실제 동작               |


```python
from typing import Tuple, List, Dict

# 타입 힌트 (설명서 역할)
def process_data(point: Tuple[int, int], values: List[int]) -> Dict[str, int]:
    return {"x": point[0], "y": point[1], "sum": sum(values)}

# 실제 데이터 (런타임 객체)
p: tuple[int, int] = (5, 7) # Python 3.9+에서는 소문자 자료형도 타입 힌트로 사용 가능
v: list[int] = [1, 2, 3, 4]
result = process_data(p, v)

print(result) # {'x': 5, 'y': 7, 'sum': 10}
```
