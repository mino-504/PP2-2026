숫자 추측 게임

```python
from tkinter import *
import random

answer = random.randint(1, 100)

def guessing():
    guess = int(guessField.get())
    
    if guess > answer:
        msg = "높음!"
    elif guess < answer:
        msg = "낮음"
    else:
        msg = "정답"
        
    resultLabel["text"] = msg
    guessField.delete(0, 5)
    
def reset():
    global answer
    answer = random.randint(1, 100)
    resultLabel["text"] = "다시 한번 하세요!"

window = Tk()
window.configure(bg="white")
window.title("숫자를 맞춰보세요!")

window.geometry("500x80")
titleLabel = Label(window, text="숫자 게임에 오신 것을 환영합니다!", bg="white")
titleLabel.pack()

guessField = Entry(window)
guessField.pack(side="left")
tryButton = Button(window, text="시도", fg="green", bg="white", command=guessing )
tryButton.pack(side="left")

resetButton = Button(window, text="초기화", fg="red", bg="white", command=reset)
resetButton.pack(side="left")
resultLabel = Label(window, text="1부터 100사이의 숫자를 입력하시오.", bg="white")
resultLabel.pack(side="left")

window.mainloop()
```

Lab: 가위, 바위, 보 게임

```python
import random
from tkinter import *

window = Tk()

Label(window, text="선택하세요", font=("Helvetica", "16")).pack()

frame = Frame(window)

rock_image = PhotoImage(file="rock.png", master=window)
paper_image = PhotoImage(file="paper.png", master=window)
scissors_image = PhotoImage(file="scissors.png", master=window)

# 교재 오타!!
def pass_s():
    decide("가위")

def pass_r():
    decide("바위")

def pass_p():
    decide("보")

rock = Button(frame, image=rock_image, command=pass_r)
rock.pack(side="left")

paper = Button(frame, image=paper_image, command=pass_p)
paper.pack(side="left")

scissors = Button(frame, image=scissors_image, command=pass_s)
scissors.pack(side="left")

frame.pack()

Label(window, text="컴퓨터는 다음을 선택하였습니다.", font=("Helvetica", "16")).pack()

computer_image = Label(window, image=rock_image)
computer_image.pack()

output = Label(window, text="", font=("Helvetica", "16"))
output.pack()

def decide(human):
    computer = random.choice(["가위", "바위", "보"])

    if computer == "바위":
        computer_image["image"] = rock_image
    elif computer == "보":
        computer_image["image"] = paper_image
    else:
        computer_image["image"] = scissors_image

    if ((computer == "바위" and human == "보") or
        (computer == "보" and human == "가위") or
        (computer == "가위" and human == "바위")):
        result = "인간 승리!"
    elif computer == human:
        result = "비겼습니다."
    else:
        result = "컴퓨터 승리!"

    output.config(text="인간: " + human + " 컴퓨터: " + computer + " " + result)

window.mainloop()
```

```python
from tkinter import *

window = Tk()
window.title("My Calculator")
display = Entry(window, width=33, bg="yellow")
display.grid(row=0, column=0, columnspan=5)

button_list = [
'7',  '8',  '9',  '/',  'C',
'4',  '5',  '6',  '*',  ' ',
'1',  '2',  '3',  '-',  ' ',
'0',  '.',  '=',  '+',  ' ' ]

def click(key):
    if   key == "=":
        result = eval(display.get())
        s = str(result)
        display.insert(END, "=" + s)
    else:
        display.insert(END, key)

row_index = 1
col_index = 0
for button_text in button_list:
    Button(window, text=button_text, width=5, 	
		command=lambda t=button_text: click(t)).grid(row=row_index, column=col_index)
    col_index += 1
    if   col_index > 4:
        row_index += 1
        col_index = 0
window.mainloop()
```

Lab: 그림판 프로그램 만들기

```python

from tkinter import *
from tkinter.colorchooser import askcolor

DEFAULT_PEN_SIZE = 3
DEFAULT_COLOR = "black"

mode = "pen"
old_x = None
old_y = None
mycolor = DEFAULT_COLOR

# 펜 선택
def use_pen():
    global mode
    mode = "pen"

# 브러쉬 선택
def use_brush():
    global mode
    mode = "brush"

# 색상 선택
def choose_color():
    global mycolor, mode

    color = askcolor(color=mycolor, title="색상 선택")[1]

    if color is not None:
        mycolor = color
        mode = "pen"
        colorButton.config(bg=mycolor)

# 지우개 선택
def use_eraser():
    global mode
    mode = "erase"

# 도구별 색상과 굵기 설정
def get_pen_settings():
    size = var.get()
    if mode == "erase":
        return "white", size * 4
    elif mode == "brush":
        return mycolor, size * 3
    else:
        return mycolor, size

# 마우스 클릭 시 점 찍기
def start_paint(event):
    global old_x, old_y

    old_x = event.x
    old_y = event.y

    color, size = get_pen_settings()
    radius = size / 2

    canvas.create_oval(
        event.x - radius,
        event.y - radius,
        event.x + radius,
        event.y + radius,
        fill=color,
        outline=color
    )

# 마우스 드래그 시 그림 그리기
def paint(event):
    global old_x, old_y

    if old_x is None or old_y is None:
        start_paint(event)
        return

    color, size = get_pen_settings()

    canvas.create_line(
        old_x, old_y,
        event.x, event.y,
        width=size,
        fill=color,
        capstyle=ROUND
    )

    old_x = event.x
    old_y = event.y

# 마우스 버튼을 떼면 좌표 초기화
def reset(event=None):
    global old_x, old_y
    old_x = None
    old_y = None

# 모두 삭제
def clear_all():
    canvas.delete(ALL)
    reset()

# 메인 윈도우
window = Tk()
window.title("간단한 그림판")

# 슬라이더 변수
var = IntVar(value=DEFAULT_PEN_SIZE)

# 버튼 생성
penButton = Button(window, text="펜", command=use_pen)
penButton.grid(row=0, column=0, sticky=W+E)

brushButton = Button(window, text="브러쉬", command=use_brush)
brushButton.grid(row=0, column=1, sticky=W+E)

colorButton = Button(window, text="색상선택", command=choose_color)
colorButton.grid(row=0, column=2, sticky=W+E)

eraseButton = Button(window, text="지우개", command=use_eraser)
eraseButton.grid(row=0, column=3, sticky=W+E)

clearButton = Button(window, text="모두삭제", command=clear_all)
clearButton.grid(row=0, column=4, sticky=W+E)

# 굵기 조절 슬라이더
scale = Scale(
    window,
    variable=var,
    from_=1,
    to=20,
    orient=VERTICAL,
    label="굵기"
)
scale.grid(row=1, column=5, sticky=N+S)

# 그림판 캔버스
canvas = Canvas(window, bg="white", width=600, height=400)
canvas.grid(row=1, column=0, columnspan=5)

# 마우스 이벤트 연결
canvas.bind("<Button-1>", start_paint)
canvas.bind("<B1-Motion>", paint)
canvas.bind("<ButtonRelease-1>", reset)

window.mainloop()
```

Lab: 공 애니메이션1

```python

from tkinter import *
import random

window = Tk()
window.title("움직이는 공")

canvas = Canvas(window, width=600, height=400)
canvas.pack()

class Ball:
    def __init__(self, color, size):
        self.id = canvas.create_oval(0, 0, size, size, fill=color)
        self.dx = random.randint(1, 10)
        self.dy = random.randint(1, 10)

    def move(self):
        canvas.move(self.id, self.dx, self.dy)

        x0, y0, x1, y1 = canvas.coords(self.id)

        if y1 >= canvas.winfo_height() or y0 <= 0:
            self.dy = -self.dy

        if x1 >= canvas.winfo_width() or x0 <= 0:
            self.dx = -self.dx

ball1 = Ball("blue", 60)
ball2 = Ball("green", 100)
ball3 = Ball("orange", 80)

def animate():
    ball1.move()
    ball2.move()
    ball3.move()
    window.after(50, animate)
    
animate()

window.mainloop()
```

Lab: 공 애니메이션2


```python

from tkinter import *
import random

window = Tk()

canvas = Canvas(window, width=600, height=400)
canvas.pack()

class Ball:
    def __init__(self, color, size):
        self.id = canvas.create_oval(0, 0, size, size, fill=color)
        self.dx = random.randint(1, 10)
        self.dy = random.randint(1, 10)

    def move(self):
        canvas.move(self.id, self.dx, self.dy)
        x0, y0, x1, y1 = canvas.coords(self.id)

        if y1 > canvas.winfo_height() or y0 < 0:
            self.dy = -self.dy

        if x1 > canvas.winfo_width() or x0 < 0:
            self.dx = -self.dx

colors = ["red", "orange", "yellow", "green", "blue", "indigo", "violet"]

ballList = []

for i in range(30):
    ballList.append(Ball(random.choice(colors), 60))

def animate():
    for i in range(30):
        ballList[i].move()

    window.after(50, animate)

animate()

window.mainloop()
```
