# ipynb 단축키 목록

| 단축키             | 기능                 |
| --------------- | ------------------ |
| `Enter`         | 선택한 셀 편집 모드        |
| `Esc`           | 명령 모드로 전환          |
| `Shift + Enter` | 현재 셀 실행 후 다음 셀로 이동 |
| Ctrl + Enter    | 현재 셀만 실행           |
| `A`             | 현재 셀 위에 새 셀 추가     |
| `B`             | 현재 셀 아래에 새 셀 추가    |
| `D D`           | 현재 셀 삭제            |
| `Z`             | 삭제한 셀 복구           |
| `M`             | 셀을 Markdown으로 변경   |
| `Y`             | 셀을 Code로 변경        |
| C               | 셀 복사               |
| X               | 셀 잘라내기             |
| V               | 아래에 붙여넣기           |
| Shift + V       | 위에 붙여넣기            |
| P               | 명령 팔레트 열기          |
| L               | 코드 셀 줄 번호 표시 / 숨김  |
