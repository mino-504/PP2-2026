# 상속

- 상속에 대한 이론과 실제 학습

```python
class Car:
    def __init__(self, maker, model, color, price):
        self.maker = maker
        self.model = model
        self.color = color
        self.price = price

    def setMake(self, maker):
        self.maker = maker

    def getMaker(self):
        return self.maker

    def getDesc(self):
        return self.__str__()

    def __str__(self):
        return f"Maker: {self.maker}, Model: {self.model}, Color: {self.color}, Price: {self.price}"


class EletricCar(Car):
    def __init__(self, maker, model, color, price, bat_size):
        super().__init__(maker, model, color, price)
        self.batterySize = bat_size

    def getBatterySize(self):
        return self.batterySize

    def setBatterySize(self, bat_size):
        self.batterySize = bat_size

    def __str__(self):
        return f"{super().__str__()}, Battery : {self.batterySize}kmh"

    def getDesc(self):
        return self.__str__()


class Truck(Car):
    def __init__(self, maker, model, color, price, payLoad):
        super().__init__(maker, model, color, price)
        self.payLoad = payLoad

    def __str__(self):
        return f"{super().__str__()}, PayLoad : {self.payLoad}kg"


car1 = Car("Kia", "Ray", "Blue", 2000)
print(car1.getDesc())
print(car1.model)

ecar1 = EletricCar("Tesla", "Model S", "Red", 8000, 81)
print(ecar1)
ecar1.setBatterySize(92)
print(ecar1.getDesc())

truc1 = Truck("Ford", "F-150", "Black", 4000, 1000)
print(truc1)
print(truc1.getMaker())

print(type(truc1))
print(type(ecar1))
print(type(car1))

print(isinstance(truc1, Truck))
print(isinstance(ecar1, EletricCar))
print(isinstance(car1, Car))

print(isinstance(truc1, Car))
print(isinstance(ecar1, Car))

#print(isinstance(truc1, Vehicle))
```
