# 마크다운문법 정리

| 기능 | 작성 방법 | 설명 |
|---|---|---|
| 제목 1 | # 제목 | 가장 큰 제목 |
| 제목 2 | ## 제목 | 중간 제목 |
| 제목 3 | ### 제목 | 작은 제목 |
| 굵게 | **문자** | 글자를 굵게 표시 |
| 기울임 | *문자* | 글자를 기울여 표시 |
| 취소선 | ~~문자~~ | 글자에 취소선 표시 |
| 인라인 코드 | `print()` | 짧은 코드 표시 |
| 글머리표 | - 항목 | 순서 없는 목록 |
| 번호 목록 | 1. 항목 | 순서 있는 목록 |
| 인용문 | > 내용 | 인용문 표시 |
| 구분선 | --- | 가로 구분선 |
| 링크 | [이름](주소) | 링크 삽입 |
| 이미지 | ![설명](이미지주소) | 이미지 삽입 |
| 표 | \| 제목 \| 내용 \| | 표 작성 |
| 줄바꿈 | 문장 끝 공백 2칸 | 다음 줄로 이동 |
| 코드 블록 | ```python 코드 ``` | 여러 줄 코드 표시 |
